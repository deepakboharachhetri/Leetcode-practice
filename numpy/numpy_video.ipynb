List vs numpy array 

````python
list1 =[1,2,3]
print("python list multiplication ",list1*2)  #here  2 is constant multiplication 


array_list=np.array(list1)
print("python list multiplication ",array_list*2) #2 is  multiply by scalar , element wise multiplicaiton 
````

````python
import time 

start_time=time.time()
list2= [i*2 for i in range(100000)]
print("total time taken",time.time()-start_time)


start_time=time.time()
list2= np.arange(100000)*2
print("total time taken",time.time()-start_time)

````
